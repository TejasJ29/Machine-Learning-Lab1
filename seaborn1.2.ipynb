```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
sns.set_style("darkgrid")

plt.figure(figsize=(8,6))
sns.scatterplot(data=titanic, x="age", y="fare", hue="survived", style="sex", size="pclass")
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare in $")
plt.show()
```

```python
sns.boxplot(data=titanic, x="class", y="fare", hue="sex")
plt.title("Fare by Class and Gender")
plt.show()
```

```python
sns.barplot(data=titanic, x="class", y="survived", estimator="mean")
plt.title("Survival Rate by Class")
plt.show()
```
