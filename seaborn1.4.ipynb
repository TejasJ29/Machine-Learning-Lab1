```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
from sklearn.datasets import fetch_openml

house_boston = fetch_openml(name="Boston", version=1, as_frame=True)
house = house_boston.data.copy()
house['PRICE'] = house_boston.target
house.head()
print(house_boston.DESCR)
```

```python
sns.histplot(house['PRICE'])
```

```python
correlation = house.corr()
plt.figure(figsize=(10,10))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f", cbar=True)
plt.title("Correlation Matrix of Boston Housing Features")
plt.show()
```
