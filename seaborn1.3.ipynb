```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
planets = sns.load_dataset("planets")
planets.head()
```

```python
top_methods = planets["method"].value_counts().nlargest(4).index
planets_top = planets[planets["method"].isin(top_methods)]
```

```python
sns.set_style("darkgrid")

plt.figure(figsize=(8,6))
sns.scatterplot(data=planets_top, x="distance", y="mass", hue="method", size="number")
plt.title("Distance vs Mass")
plt.xlabel("Distance (parsecs)")
plt.ylabel("Mass (Jupiter masses)")
plt.show()
```

```python
sns.boxplot(data=planets_top, x="method", y="orbital_period")
plt.yscale("log")
plt.title("Orbital Period by Detection Method")
plt.ylabel("Orbital Period (days, log scale)")
plt.show()
```

```python
sns.barplot(data=planets_top, x="method", y="mass", estimator="mean")
plt.title("Average Mass by Detection Method")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(data=planets_top, x="method")
plt.title("Number of Planets Found by Detection Method")
plt.xlabel("Detection Method")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,6))
discoveries_per_year = planets.groupby("year").size().reset_index(name="count")
sns.lineplot(data=discoveries_per_year, x="year", y="count", marker="o")
plt.title("Planet Discoveries per Year")
plt.xlabel("Year")
plt.ylabel("Number of Planets Discovered")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(planets.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Between Numeric Columns")
plt.show()
```
